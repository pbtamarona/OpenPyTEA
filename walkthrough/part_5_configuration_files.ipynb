## 📁 TEA Using Input and Output Configuration Files

In addition to defining equipment and plant configurations directly in Python, OpenPyTEA supports a configuration-based workflow through structured JSON input and output files, handled via functions in the `io.py` module. This approach enables more reproducible and scalable analyses, particularly for larger studies, batch evaluations, or integration with external process simulation tools.

In this workflow, equipment, plant parameters, and analysis settings are defined in separate configuration files. These inputs are then processed to automatically run the techno-economic analysis and generate structured outputs, including metrics, breakdown data, and plots.

To see how this workflow is implemented in practice, please refer to the Case Study 1 notebook: [`examples/case_study_1_with_json.ipynb` ](https://github.com/pbtamarona/OpenPyTEA/blob/main/examples/case_study_1_with_JSON.ipynb) 

Installing **OpenPyTEA** also installs an `openpytea` command-line tool, so this same JSON-based workflow can be run directly from a terminal without writing any Python. The single-file variant (equipment + plant + analysis combined in one JSON file) is run with:

```bash
openpytea run project/config.json --output-dir outputs/tea_results
```

`openpytea equipment`, `openpytea plant`, and `openpytea tea` (the three-file variant, matching separate equipment/plant/analysis configs) are also available — run `openpytea --help` for the full command list.
